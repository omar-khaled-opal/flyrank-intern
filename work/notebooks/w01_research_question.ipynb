## 1. My lane (or freestyle) and why

**Research question:** Which content items should be prioritized for review or refresh based on their observed freshness and search-performance signals?

**Decision:** Decide which content items are more worth prioritizing for a content review or refresh.

**Action:** A content or SEO team could review the prioritized items and decide whether to update, improve, or leave them unchanged.

**Cost of a wrong recommendation:** Prioritizing the wrong content could waste limited content and SEO resources. Missing a useful refresh opportunity could also mean leaving potential search performance improvements unexplored. Therefore, any result should be treated as decision support rather than a guarantee that refreshing a page will improve its performance.


```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Average CTR:", df["ctr"].mean())
print("Average position:", df["avg_position"].mean())
print("Average content age:", df["content_age_days"].mean())
```

The starter dataset contains 30,000 content items. The measured average CTR is about 0.51%, while the average search position is about 16.34. The average content age is about 256 days. These numbers show that the dataset contains a substantial number of content observations together with measurable search-performance and freshness signals, making it reasonable to investigate whether these signals can support content-refresh prioritization.
